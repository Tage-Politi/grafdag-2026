```python
import time
import networkx as nx
from ipysigma import Sigma
```

```python
import gzip

G = nx.DiGraph()
fd = gzip.open("data/email.edgelist.txt.gz", "rt")
for linje in fd:
    s, m = linje.split()
    s = int(s)
    m = int(m)
    G.add_edge(s, m)
#
G.remove_edges_from(nx.selfloop_edges(G))
print(f"Nodes: {G.number_of_nodes()}")
print(f"Edges: {G.number_of_edges()}")
```

```python
print(list(G.adj[63]))
print(list(G.adj[407]))
```

```python
import math
log_degrees = {node: math.log(deg + 1) for node, deg in G.degree()}
Sigma(
    G, 
    node_size=log_degrees,
    node_color=log_degrees,
    node_color_gradient="Plasma",
    node_size_range=(1, 30),
    default_edge_type="curve"
)
```
